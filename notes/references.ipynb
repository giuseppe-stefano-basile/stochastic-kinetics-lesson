# References and attribution

- N. G. van Kampen, *Stochastic Processes in Physics and Chemistry*, 3rd ed., North-Holland (2007). Chapters III–IV introduce the hierarchy of distributions and the Markov property in the notation used in the [foundations notes](00_foundations.ipynb). Chapters VIII and XII cover the Fokker–Planck equation and first passage.
- L. Sagresti, L. Peri, G. Ceccarelli, G. Brancato, “Stochastic Model of Solvent Exchange in the First Coordination Shell of Aqua Ions,” *Journal of Chemical Theory and Computation* **18**, 3164–3173 (2022), [DOI: 10.1021/acs.jctc.2c00181](https://doi.org/10.1021/acs.jctc.2c00181). Figures reproduced in the lecture slides are credited in the [slide build notes](../slides/SNS_BUILD.md).
- L. Sagresti, [HandsOnSession_CMLS](https://github.com/lianctrl/HandsOnSession_CMLS), teaching repository, especially `MSM-FP/HandsOn_MSM.ipynb`. Inspiration for the broad teaching structure; no original code or data copied. Consult its license before reuse.
- P. Hänggi, P. Talkner, M. Borkovec, “Reaction-rate theory: fifty years after Kramers”, *Reviews of Modern Physics* **62**, 251–341 (1990), [doi:10.1103/RevModPhys.62.251](https://doi.org/10.1103/RevModPhys.62.251). Rate theory and its assumptions.
- G. Hummer, “Position-dependent diffusion coefficients and free energies from Bayesian analysis of equilibrium and replica molecular dynamics simulations”, *New Journal of Physics* **7**, 34 (2005), [doi:10.1088/1367-2630/7/1/034](https://doi.org/10.1088/1367-2630/7/1/034). - S. Redner, *A Guide to First-Passage Processes*, Cambridge University Press (2001), [doi:10.1017/CBO9780511606014](https://doi.org/10.1017/CBO9780511606014). First-passage background.
