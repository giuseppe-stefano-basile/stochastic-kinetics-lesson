## 1. Probability and a stochastic trajectory

A process $(X_t)_{t\ge 0}$ is a family of random variables indexed by time. A trajectory is one realization across time. The transition density $p(x,t\mid x_0,0)$ describes an ensemble conditional on $X_0=x_0$. In van Kampen's notation it is $P_{1|1}(x,t\mid x_0,0)$.

For Brownian motion, increments over disjoint intervals are independent Gaussian variables with mean zero and variance equal to the interval length. Thus $\Delta W\sim\sqrt{\Delta t}\,\xi$ in distribution, with $\xi\sim\mathcal N(0,1)$. The squared increment has expectation $\Delta t$ and contributes at that order to Itô's formula. Brownian paths are almost surely nowhere differentiable, so white noise is a distributional derivative.

## 2. Itô generator and forward equation

Consider the Itô diffusion $dX_t=a(X_t)dt+\sqrt{2D(X_t)}\,dW_t$, where $D>0$. For a smooth test function $f$, Itô's formula yields

$$df(X_t)=(af'+Df'')(X_t)dt+\sqrt{2D(X_t)}\,f'(X_t)dW_t.$$

The stochastic integral has mean zero under suitable integrability. The backward generator is $\mathcal L f=af'+Df''$, and $d\mathbb E[f(X_t)]/dt=\mathbb E[(\mathcal L f)(X_t)]$. Writing this expectation as an integral against $p$ and integrating by parts gives

$$\partial_t p=\mathcal L^*p=-\partial_x(ap)+\partial_x^2(Dp)=-\partial_xJ,
\qquad J=ap-\partial_x(Dp).$$

On a finite interval the boundary term in $\int (\mathcal L f)p-\int f\mathcal L^*p$ is $[fJ+Dp f']$ evaluated at the endpoints. Reflection pairs $J=0$ with $f'=0$. Killing at a regular absorbing target pairs $p=0$ with $f=0$. The forward operator evolves densities in the arrival coordinate. The backward generator acts on observables as functions of the starting coordinate.

## 3. Equilibrium and variable diffusion

At temperature $T$, write $\beta=(k_BT)^{-1}$ in energy per particle units. For constant $D$ and $a=-D\beta F'$,

$$J=-D(\beta F'p+p'),\qquad \partial_t p=-\partial_xJ.$$

The normalizable zero-current density is $p_{\rm eq}=Z^{-1}e^{-\beta F}$. Stationarity on a periodic domain can also allow a nonzero current. Here we assume equilibrium detailed balance. At fixed $F$, multiplying a constant $D$ by $c>0$ multiplies the generator by $c$ and divides MFPTs by $c$.

For a specified $D(x)$, imposing $J=0$ with the same equilibrium density gives $a=D'-D\beta F'$. The equilibrium Itô model is

$$dX_t=[D'-D\beta F'](X_t)dt+\sqrt{2D(X_t)}\,dW_t,
\qquad \partial_t p=\partial_x\left[D e^{-\beta F}\partial_x(e^{\beta F}p)\right].$$

The drift depends on the stochastic convention. For example, the equivalent Stratonovich drift subtracts $D'/2$ from the Itô drift. The units are $[D]=[x]^2/[t]$ and $[\beta F]=1$. With an invertible smooth coordinate change $y=g(x)$ and $g'\ne0$, density conservation and Itô's formula give $F_y=F_x+k_BT\ln|g'|+C$ and $D_y=(g')^2D_x$.

## 4. Reflecting and absorbing boundaries

In this and the following sections, $a$ denotes the left endpoint. For reflection at $a$ and absorption at $b$, the killed density obeys $J(a,t)=0$ and $p(b,t)=0$. Its mass is the survival probability:

$$S_x(t)=\Pr_x(\tau_b>t)=\int_a^b p(y,t\mid x,0)\,dy,
\qquad \frac{dS_x}{dt}=J(a,t)-J(b,t)=-J(b,t).$$

Thus $-S_x'(t)$ is the first-passage density where it exists. The corresponding backward conditions are $T'(a)=0$ and $T(b)=0$. Reflecting overshoots and checking the target at step endpoints are numerical approximations. Paths can hit the target and return between sampled endpoints.

## 5. Backward first-passage equation and integral

Define $\tau_b=\inf\{t\ge0:X_t\ge b\}$ and $T(x)=\mathbb E_x[\tau_b]$, assumed finite. With $\sigma=\min(\Delta t,\tau_b)$, the Markov property gives $T(x)=\mathbb E_x[\sigma+T(X_\sigma)]$. Taking the interior small-time limit, or using the stopped Dynkin formula, gives $\mathcal L T=-1$. In divergence form,

$$\mathcal L T=e^{\beta F}\frac{d}{dx}\left[D e^{-\beta F}T'\right]=-1.$$

Multiply by $e^{-\beta F}$ and integrate from $a$ to $y$. Using $T'(a)=0$ gives

$$D(y)e^{-\beta F(y)}T'(y)=-\int_a^y e^{-\beta F(z)}\,dz.$$

Integrating from $x$ to $b$ and applying $T(b)=0$ gives

$$T(x)=\int_x^b\frac{e^{\beta F(y)}}{D(y)}
\left[\int_a^y e^{-\beta F(z)}\,dz\right]dy.$$

With $F=0$ and constant $D$, this becomes $[(b-a)^2-(x-a)^2]/(2D)$. From the survival curve one also has $\mathbb E[\tau_b]=\int_0^\infty S(t)dt$, allowing an infinite value. An integral only up to $T_{\rm obs}$ gives the restricted mean $\mathbb E[\min(\tau_b,T_{\rm obs})]$. A fixed start, equilibrium within a basin, and a quasi-stationary preparation can give different mean times.

## 6. Committor and reaction rates

With both endpoints absorbing, the forward committor is $q(x)=\Pr_x(\tau_b<\tau_a)$. It solves $\mathcal L q=0$ with $q(a)=0$ and $q(b)=1$. With variable $D$,

$$q(x)=\frac{\int_a^x e^{\beta F(y)}/D(y)\,dy}
{\int_a^b e^{\beta F(y)}/D(y)\,dy}.$$

$1/T(x)$ is the reciprocal mean of the specified first-passage experiment. Interpreting it as a two-state rate requires suitable state definitions, initial preparation and timescale separation. First arrival at the barrier top and first arrival in the product basin are distinct events.

## 7. Grid generator and MSM connection

For uniform nodes $x_i$ with spacing $h$, the teaching code uses nearest-neighbour rates at constant $D$,

$$Q_{i,i+1}=\frac{D}{h^2}e^{-\beta(F_{i+1}-F_i)/2},\qquad
Q_{i+1,i}=\frac{D}{h^2}e^{+\beta(F_{i+1}-F_i)/2}.$$

Omit jumps across reflecting endpoints and set $Q_{ii}=-\sum_{j\ne i}Q_{ij}$. Rows sum to zero, off-diagonals are nonnegative, and $\pi_i\propto e^{-\beta F_i}$ satisfies detailed balance $\pi_i Q_{ij}=\pi_j Q_{ji}$. A row distribution evolves as $\dot p=pQ$ with transition matrix $P(t)=e^{tQ}$. The same $Q$ acts on an observable column $f$.

Let $S$ be the non-target nodes. The hitting-time solve is $-Q_{SS}\mathbf{T}_S=\mathbf{1}$ with zero time at the target. Finite grid spacing affects the boundary approximation. Compare several grids with continuum quadrature. An MSM fitted at a finite lag involves a separate approximation.

## 8. Molecular interpretation and limitations

For a collective variable $s(\mathbf R)$, the projected equilibrium density and free energy are

$$p_{\rm eq}(s)=\int d\mathbf R\,\rho_{\rm eq}(\mathbf R)\,\delta[s-s(\mathbf R)],
\qquad F(s)=-k_BT\ln p_{\rm eq}(s)+C.$$

Projection can introduce memory and position-dependent transport. A one-dimensional Markov diffusion is a modeling assumption. Check lag dependence, Chapman–Kolmogorov consistency, relaxation of hidden coordinates, first-passage predictions and sensitivity to state boundaries. Distinguish force-field error, uncertainty in $F$ and $D$, and model error. Recovering a free energy from metadynamics alone does not recover unbiased trajectory kinetics.
